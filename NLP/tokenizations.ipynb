```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
corpus="Hey, I am Sachin. I study here in SIT Tumkur in ISE branch. Currently" \
", i am a 3rd year student and learning NLP techniques."

print(corpus)
```

```python
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
for senteces in documents:
    print(senteces)
```

```python
for sentence in documents:
    print(sentence)
    for words in sentence:
        print(words)
```

```python
from nltk.tokenize import word_tokenize

vocabulary=word_tokenize(corpus)
```

```python
vocabulary
```

```python
for sentences in documents:
    print(word_tokenize(sentences))
```

#### With treebankWordTokenizer the full stop will nnot be considered as the separate word. except the last fullstop.

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
