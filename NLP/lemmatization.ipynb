### Lemmatization:
##### Instead of returning the root stem like stemming techniques (porterStemmer, RegExpStemmer, SnowballStemmer) it return the root word from the dictionary

##### We can provide the parts of speech(POS) tag in order to achieve the more accurate meaning of the word

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer=WordNetLemmatizer()
```

```python
words=['writer','writing','writes','player','playing','congratulations','sportingly','fairly']
```

```python
for word in words:
    print(f"{word}----->{lemmatizer.lemmatize(word,pos='v')}")
```
