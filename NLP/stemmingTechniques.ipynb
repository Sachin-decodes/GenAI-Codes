## Stemming :
##### Reducing the word to its base form

```python
words=['writing','writer','teaching','teacher','history','congratulations','finally','finalized']
```

#### PorterStemmer
###### There are some demerits of the porter stemmer that it changes the meaning of some of the words 

```python
from nltk.stem import PorterStemmer

stemming=PorterStemmer()
```

```python
for word in words:
    print(f"{word}---->{stemming.stem(word)}")
```

#### RegExpStemmer
#### More of like user defined stemming

```python
from nltk.stem import RegexpStemmer

reg_stem=RegexpStemmer('ing$|es$|ed$|able$',min=4)
```

```python
new_words=['eating','running','capable','reached','teaches','suspended']

for word in new_words:
    print(f"{word}---->{reg_stem.stem(word)}")
```

#### SnowballStemmer
##### It has better accuracy than the porter stemmer but it still has some deformities . So, it can't be used directly in chatbots.

```python
from nltk.stem import SnowballStemmer

snowball_stemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(f"{word}---->{snowball_stemmer.stem(word)}")
```

##### Here we can see the difference between the snowball stemmer and the porterStemmer

```python
snowball_stemmer.stem('fairly'),stemming.stem('fairly')
```
