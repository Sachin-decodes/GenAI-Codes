```python
import nltk
```

```python
from nltk.tokenize import word_tokenize
```

```python
sentence="The Eiffel Tower is one of the most famous landmarks in the world. It is located in Paris, France, and was completed in 1889. It was built by Gustave Eiffel's engineering company and is approximately 330 metres tall."
```

```python
words=word_tokenize(sentence)
words=nltk.pos_tag(words)
print(words)
```

```python
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
ner=nltk.ne_chunk(words)
#nltk.ne_chunk(words).draw()
print(ner)
```
